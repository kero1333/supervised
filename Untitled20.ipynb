```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("/content/OnlineRetail.csv", encoding = "unicode_escape")
```

```python
data = df[df['CustomerID'].notna()]
```

```python
data.isnull().sum()
```

```python
df_fix = data.sample(10000 , random_state=42)
```

```python
df_fix ["Total price"] = df_fix["Quantity"]*df_fix["UnitPrice"]
```

```python
df_fix.head()
```

```python
costumer_amount = df_fix.groupby("CustomerID")["Total price"].sum()
```

```python
costumer_amount.head()
```

```python
from sklearn.preprocessing import StandardScaler
standard_scaler = StandardScaler()
data_scaled = standard_scaler.fit_transform(costumer_amount.values.reshape(-1, 1))
```

```python
# View a sample of the scaled data
display(data_scaled[:5])
```

```python
inertia = []
for k in range(1, 10):
  kmeans = KMeans(n_clusters=k, random_state=42, n_init='auto')
  kmeans.fit(data_scaled)
  inertia.append(kmeans.inertia_)

# Plot the elbow curve
plt.figure(figsize=(8, 5))
plt.plot(range(1, 10), inertia, marker='o', linestyle='--')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal k')
plt.grid(True)
plt.show()
```
